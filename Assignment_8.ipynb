### Q1.  Write a unique paragraph (5-6 sentences) about your favorite topic (e.g., sports, technology, food, books, etc.). 
- 1. Convert text to lowercase and remove punctua on using re.
- 2. Tokenize the text into words and sentences.
- 3. Split using split() and word_tokenize() and compare how Python split and NLTK’s word_tokenize() differ.
- 4. Remove stopwords (using NLTK's stopwords list).
- 5. Display word frequency distribu on (excluding stopwords).

```python
import re
import nltk
import matplotlib.pyplot as plt
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

paragraph = """Artificial intelligence is changing the world.
Technology makes our daily life easier and faster.
Machine learning helps computers learn from data.
Students use technology to improve their knowledge.
Robots and smart devices are becoming more common.
I believe artificial intelligence has a bright future."""

print("Original Paragraph:")
print(paragraph)

# 1. Convert to lowercase and remove punctuation
text = paragraph.lower()
text = re.sub(r'[^\w\s]', '', text)

print("\nLowercase Text Without Punctuation:")
print(text)

# 2. Tokenize into words and sentences
words = word_tokenize(text)
sentences = sent_tokenize(paragraph)

print("\nWord Tokenization:")
print(words)

print("\nSentence Tokenization:")
print(sentences)

# 3. Compare split() and word_tokenize()
split_words = paragraph.split()
nltk_words = word_tokenize(paragraph)

print("\nUsing split():")
print(split_words)

print("\nUsing word_tokenize():")
print(nltk_words)

# 4. Remove stopwords
stop_words = set(stopwords.words('english'))

filtered_words = []

for word in words:
    if word not in stop_words:
        filtered_words.append(word)

print("\nWords After Removing Stopwords:")
print(filtered_words)

# 5. Word Frequency Distribution
frequency = FreqDist(filtered_words)

print("\nWord Frequency Distribution:")
print(frequency)

print("\nMost Common Words:")
print(frequency.most_common())

frequency.plot()
plt.show()
```

### Q2. Using the same paragraph from Q1: 
- 1. Extract all words with only alphabets using re.findall() 
- 2. Remove stop words using NLTK’s stopword list 
- 3. Perform stemming with PorterStemmer
- 4. Perform lemma za on with WordNetLemma zer 
- 5. Compare the stemmed and lemma zed outputs and explain when you’d prefer one over the other. 

```python
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize

nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

# Same paragraph from Q1
text = paragraph.lower()

# 1. Extract alphabetic words
words = re.findall(r'\b[a-zA-Z]+\b', text)

print("Alphabetic Words:")
print(words)

# 2. Remove stopwords
stop_words = set(stopwords.words('english'))

filtered_words = []

for word in words:
    if word not in stop_words:
        filtered_words.append(word)

print("\nAfter Removing Stopwords:")
print(filtered_words)

# 3. Stemming
stemmer = PorterStemmer()

stemmed_words = []

for word in filtered_words:
    stemmed_words.append(stemmer.stem(word))

print("\nStemmed Words:")
print(stemmed_words)

# 4. Lemmatization
lemmatizer = WordNetLemmatizer()

lemmatized_words = []

for word in filtered_words:
    lemmatized_words.append(lemmatizer.lemmatize(word))

print("\nLemmatized Words:")
print(lemmatized_words)

# 5. Comparison
print("\nComparison:")

for i in range(len(filtered_words)):
    print(filtered_words[i], "-> Stem:", stemmed_words[i],
          "| Lemma:", lemmatized_words[i])
```

### Q3. Choose 3 short texts of your own (e.g., different news headlines, product reviews). 
- 1. Use CountVectorizer to generate the Bag of Words representa on. 
- 2. Use TfidfVectorizer to compute TF-IDF scores. 
- 3. Print and interpret the top 3 keywords from each text using TF-IDF. 

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer

# Three short texts
texts = [
    "Artificial intelligence makes technology smarter and faster",
    "Machine learning helps computers learn from data",
    "Technology makes education easier and learning better"
]

# 1. Bag of Words using CountVectorizer
cv = CountVectorizer()

bow_matrix = cv.fit_transform(texts)

print("Bag of Words Matrix:")

bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=cv.get_feature_names_out()
)

print(bow_df)


# 2. TF-IDF using TfidfVectorizer
tfidf = TfidfVectorizer()

tfidf_matrix = tfidf.fit_transform(texts)

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf.get_feature_names_out()
)

print("\nTF-IDF Matrix:")
print(tfidf_df.round(3))


# 3. Print Top 3 Keywords from Each Text

feature_names = tfidf.get_feature_names_out()

print("\nTop 3 Keywords from Each Text:")

for i in range(len(texts)):

    scores = tfidf_matrix[i].toarray()[0]

    top_indices = scores.argsort()[-3:][::-1]

    print("\nText", i + 1)

    for index in top_indices:
        print(feature_names[index], ":", round(scores[index], 3))
```

### Q4. Write 2 short texts (4–6 lines each) describing two different technologies (e.g., AI vs Blockchain). 
- 1. Preprocess and tokenize both texts. 
- 2. Calculate: a. Jaccard Similarity using sets b. Cosine Similarity using TfidfVectorizer + cosine_similarity() c. Analyze which similarity metric gives better insights in your case. 

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

text1 = """
Artificial Intelligence is a technology that allows machines to think and learn.
AI uses data to make decisions and solve problems.
It is used in healthcare, education, and transportation.
AI helps automate tasks and improve efficiency.
It is becoming an important part of modern technology.
"""

text2 = """
Blockchain is a technology used to store information securely.
It maintains digital records across multiple computers.
Blockchain is commonly used in cryptocurrency and finance.
It provides transparency and security to digital transactions.
It is becoming an important part of modern technology.
"""

# Preprocessing
def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text)
    words = text.split()
    return words

words1 = preprocess(text1)
words2 = preprocess(text2)

print("Tokens of Text 1:")
print(words1)

print("\nTokens of Text 2:")
print(words2)

# Jaccard Similarity
set1 = set(words1)
set2 = set(words2)

intersection = set1.intersection(set2)
union = set1.union(set2)

jaccard = len(intersection) / len(union)

print("\nCommon Words:", intersection)
print("Jaccard Similarity:", round(jaccard, 4))

# Cosine Similarity using TF-IDF
texts = [text1, text2]

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(texts)

cosine = cosine_similarity(tfidf_matrix[0], tfidf_matrix[1])

print("Cosine Similarity:", round(cosine[0][0], 4))

# Comparison
print("\nComparison:")

if jaccard > cosine[0][0]:
    print("Jaccard Similarity gives a higher similarity score.")
elif cosine[0][0] > jaccard:
    print("Cosine Similarity gives a higher similarity score.")
else:
    print("Both metrics give the same similarity score.")
```

### Q5. Write a short review for a product or service. 
- 1. Use TextBlob or VADER to find polarity & subjectivity for each review. 
- 2. Classify reviews into Posi ve / Negative / Neutral.
- 3. Create a word cloud using the wordcloud library for all positive reviews.

```python
!pip install textblob wordcloud
```

```python
from textblob import TextBlob
from wordcloud import WordCloud
import matplotlib.pyplot as plt

reviews = [
    "This product is amazing and I absolutely love it.",
    "The quality is excellent and the product works perfectly.",
    "I am very happy with this purchase.",
    "The product is terrible and completely useless.",
    "Very poor quality and disappointing experience.",
    "The product is okay and works as expected.",
    "It is an average product with normal features."
]

positive_reviews = []

for review in reviews:

    blob = TextBlob(review)

    polarity = blob.sentiment.polarity
    subjectivity = blob.sentiment.subjectivity

    if polarity > 0.1:
        sentiment = "Positive"
        positive_reviews.append(review)

    elif polarity < -0.1:
        sentiment = "Negative"

    else:
        sentiment = "Neutral"

    print("\nReview:", review)
    print("Polarity:", round(polarity, 2))
    print("Subjectivity:", round(subjectivity, 2))
    print("Sentiment:", sentiment)

# Create Word Cloud using positive reviews

positive_text = " ".join(positive_reviews)

wordcloud = WordCloud(
    width=800,
    height=400,
    background_color="white"
).generate(positive_text)

plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.title("Word Cloud of Positive Reviews")
plt.show()
```

### Q6. Choose your own paragraph (~100 words) as training data. 
- 1. Tokenize text using Tokenizer() from keras.preprocessing.text 
- 2. Create input sequences and build a simple LSTM or Dense model 
- 3. Train the model and generate 2–3 new lines of text star ng from any seed word you provide.

```python
import numpy as np
import tensorflow as tf

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
from tensorflow.keras.utils import to_categorical
```

```python
text = """
Technology is changing the world in many different ways.
Artificial intelligence helps machines learn from data and solve problems.
Machine learning is used in healthcare, education, and business.
Students use technology to improve their knowledge and skills.
Computers help people perform difficult tasks quickly and efficiently.
Modern technology makes communication easier and faster.
Artificial intelligence is becoming an important part of our daily lives.
Robots and smart devices are helping humans in many fields.
Learning about technology helps students prepare for the future.
Technology continues to create new opportunities for innovation.
Artificial intelligence and machine learning are transforming modern industries.
Technology allows people to connect and share information across the world.
Education has become more accessible through online learning platforms.
Students can use computers to develop new skills and knowledge.
The future of technology will bring many exciting opportunities.
"""
```

```python
tokenizer = Tokenizer()

tokenizer.fit_on_texts([text])

total_words = len(tokenizer.word_index) + 1

print("Total unique words:", total_words)

print("\nWord Index:")
print(tokenizer.word_index)
```

```python
input_sequences = []

for line in text.split('.'):
    line = line.strip()

    if line:
        token_list = tokenizer.texts_to_sequences([line])[0]

        for i in range(1, len(token_list)):
            n_gram_sequence = token_list[:i+1]

            input_sequences.append(n_gram_sequence)

print("Total input sequences:", len(input_sequences))

print("\nFirst 10 sequences:")
print(input_sequences[:10])
```

```python
max_sequence_len = max([len(x) for x in input_sequences])

input_sequences = np.array(
    pad_sequences(
        input_sequences,
        maxlen=max_sequence_len,
        padding='pre'
    )
)

print("Maximum Sequence Length:", max_sequence_len)

print("\nPadded Sequences:")
print(input_sequences[:5])
```

```python
X = input_sequences[:, :-1]

y = input_sequences[:, -1]

y = to_categorical(y, num_classes=total_words)

print("X Shape:", X.shape)
print("Y Shape:", y.shape)
```

```python
model = Sequential()

model.add(
    Embedding(
        input_dim=total_words,
        output_dim=64
    )
)

model.add(LSTM(100))

model.add(
    Dense(
        total_words,
        activation='softmax'
    )
)

model.compile(
    loss='categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

model.summary()
```

```python
history = model.fit(
    X,
    y,
    epochs=150,
    verbose=1
)
```

```python
def generate_text(seed_text, next_words):

    for _ in range(next_words):

        token_list = tokenizer.texts_to_sequences([seed_text])[0]

        token_list = pad_sequences(
            [token_list],
            maxlen=max_sequence_len - 1,
            padding='pre'
        )

        predicted = model.predict(token_list, verbose=0)

        predicted_word_index = np.argmax(predicted, axis=1)[0]

        output_word = ""

        for word, index in tokenizer.word_index.items():

            if index == predicted_word_index:
                output_word = word
                break

        if output_word == "":
            break

        seed_text += " " + output_word

    return seed_text


print("Generated Text 1:")
print(generate_text("technology", 10))

print("\nGenerated Text 2:")
print(generate_text("artificial", 10))

print("\nGenerated Text 3:")
print(generate_text("students", 10))
```
