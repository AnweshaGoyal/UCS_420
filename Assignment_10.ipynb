```python
!pip install nltk

import nltk
import re

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

print("NLTK setup completed!")
```

```python
stop_words = set(stopwords.words('english'))

lemmatizer = WordNetLemmatizer()
```

```python
def preprocess_text(text):

    # 1. Lowercase conversion
    text = text.lower()

    # Remove special characters and numbers
    text = re.sub(r'[^a-z\s]', '', text)

    # 2. Tokenization
    tokens = word_tokenize(text)

    # 3. Stopword removal
    tokens = [
        word for word in tokens
        if word not in stop_words
    ]

    # 4. Lemmatization
    tokens = [
        lemmatizer.lemmatize(word)
        for word in tokens
    ]

    return " ".join(tokens)
```

```python
df["Processed_Query"] = df["Query"].apply(preprocess_text)

display(df.head(15))
```

```python
for i in range(5):

    print("Original:", df["Query"][i])

    print("Processed:", df["Processed_Query"][i])

    print("-" * 60)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
X = df["Processed_Query"]
y = df["Intent"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
tfidf = TfidfVectorizer(
    ngram_range=(1, 2)
)

X_train_tfidf = tfidf.fit_transform(X_train)

X_test_tfidf = tfidf.transform(X_test)

print("TF-IDF feature extraction completed!")
print("Training matrix shape:", X_train_tfidf.shape)
print("Testing matrix shape:", X_test_tfidf.shape)
```

```python
nb_model = MultinomialNB()

nb_model.fit(X_train_tfidf, y_train)

nb_predictions = nb_model.predict(X_test_tfidf)

nb_accuracy = accuracy_score(y_test, nb_predictions)

print("Naive Bayes Accuracy:", nb_accuracy)
```

```python
lr_model = LogisticRegression(
    max_iter=2000,
    random_state=42
)

lr_model.fit(X_train_tfidf, y_train)

lr_predictions = lr_model.predict(X_test_tfidf)

lr_accuracy = accuracy_score(y_test, lr_predictions)

print("Logistic Regression Accuracy:", lr_accuracy)
```

```python
results = pd.DataFrame({
    "Model": ["Naive Bayes", "Logistic Regression"],
    "Accuracy": [nb_accuracy, lr_accuracy]
})

display(results)
```

```python
if lr_accuracy >= nb_accuracy:
    best_model = lr_model
    print("Selected Model: Logistic Regression")
else:
    best_model = nb_model
    print("Selected Model: Naive Bayes")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)
import matplotlib.pyplot as plt
```

```python
# Predictions using the selected model

y_pred = best_model.predict(X_test_tfidf)

accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(
    y_test, y_pred, average='weighted', zero_division=0
)

recall = recall_score(
    y_test, y_pred, average='weighted', zero_division=0
)

f1 = f1_score(
    y_test, y_pred, average='weighted', zero_division=0
)

print("Model Evaluation Results")
print("------------------------")
print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("F1 Score:", f1)
```

```python
print(classification_report(
    y_test,
    y_pred,
    zero_division=0
))
```

```python
cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=best_model.classes_
)

fig, ax = plt.subplots(figsize=(10, 8))

disp.plot(
    cmap='Blues',
    xticks_rotation=45,
    ax=ax
)

plt.title("DiabetesCare AI - Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
def rule_based_reasoning(query, intent, confidence):

    query = query.lower()

    emergency_keywords = [
        "unconscious", "faint", "seizure", "unresponsive",
        "cannot wake", "difficulty breathing"
    ]

    # Rule 1: Emergency Detection
    if any(word in query for word in emergency_keywords):
        return """EMERGENCY ALERT!
This may be a medical emergency.
Call emergency services immediately.
Do not give food or liquids to an unconscious person."""

    # Rule 2: Low Blood Sugar Detection
    if any(word in query for word in ["low blood sugar", "hypoglycemia"]):
        return """Possible low blood sugar (hypoglycemia).
If the person is conscious and able to swallow,
follow their prescribed hypoglycemia treatment plan.
Severe symptoms require emergency medical attention."""

    # Rule 3: High Blood Sugar Detection
    if any(word in query for word in ["high blood sugar", "hyperglycemia"]):
        return """High blood sugar (hyperglycemia) may require monitoring
and medical advice. Follow your prescribed diabetes
management plan. Seek urgent medical care if severe symptoms occur."""

    # Rule 4: Low Confidence Handling
    if confidence < 0.15:
        return """I could not confidently understand your question.
Please rephrase it using different words."""

    # Rule 5: Medication Safety
    if intent == "medication":
        return """Medication Alert:
Always follow your doctor's prescription.
Never change or stop diabetes medication without
consulting a healthcare professional."""

    # Rule 6: Unsupported Query Handling
    if intent not in knowledge_base:
        return """Sorry, I can only answer diabetes-related healthcare questions."""

    return None
```

```python
def diabetes_assistant(user_query):

    query_lower = user_query.lower()

    # Step 1: Emergency detection (always checked first)
    emergency_keywords = [
        "unconscious", "faint", "seizure",
        "unresponsive", "cannot wake", "difficulty breathing"
    ]

    if any(word in query_lower for word in emergency_keywords):
        return """EMERGENCY ALERT!
This may be a medical emergency.
Call emergency services immediately.
Do not give food or liquids to an unconscious person."""

    # Step 2: Unsupported query detection
    diabetes_keywords = [
        "diabetes", "diabetic", "blood sugar", "glucose",
        "insulin", "hba1c", "hypoglycemia", "hyperglycemia",
        "metformin", "pancreas", "sugar level"
    ]

    if not any(word in query_lower for word in diabetes_keywords):
        return "Sorry, I can only answer diabetes-related healthcare questions."

    # Step 3: Daily diabetes management
    management_keywords = [
        "manage my diabetes daily",
        "manage diabetes daily",
        "daily diabetes management",
        "manage diabetes"
    ]

    if any(word in query_lower for word in management_keywords):
        return """Daily Diabetes Management:
- Monitor blood glucose as recommended by your healthcare professional.
- Follow your prescribed medication schedule.
- Eat a balanced and healthy diet.
- Include regular physical activity.
- Attend regular medical checkups.
- Monitor your HbA1c as advised by your doctor.

Consult your healthcare professional for a management plan
suited to your individual needs."""

    # Step 4: NLP preprocessing
    processed_query = preprocess_text(user_query)

    # Step 5: TF-IDF transformation
    query_vector = tfidf.transform([processed_query])

    # Step 6: Intent prediction
    probabilities = best_model.predict_proba(query_vector)[0]
    predicted_intent = best_model.classes_[probabilities.argmax()]
    confidence = probabilities.max()

    # Step 7: Rule-based reasoning
    rule_response = rule_based_reasoning(
        user_query, predicted_intent, confidence
    )

    if rule_response is not None:
        return rule_response

    # Step 8: Confidence check
    if confidence < 0.15:
        return "I could not confidently understand your question. Please rephrase it."

    # Step 9: Knowledge retrieval
    information = retrieve_knowledge(predicted_intent)

    return (
        f"Detected Intent: {predicted_intent}\n"
        f"Confidence: {confidence:.2f}\n\n"
        f"Answer: {information}"
    )


# Test the updated assistant
test_queries = [
    "How can I manage my diabetes daily?",
    "What is the capital of France?",
    "What are the symptoms of diabetes?",
    "What should I do if a diabetic person faints?"
]

for query in test_queries:
    print("\nUser:", query)
    print("Assistant:", diabetes_assistant(query))
    print("-" * 60)
```
