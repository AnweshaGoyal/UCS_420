### Q2.  Questions on Basic NumPy Array:

#### a) Reverse the NumPy array: arr = np.array([1, 2, 3, 6, 4, 5])

```python
arr = np.array([1,2,3,6,4,5])
reverse = arr[::-1]
print("Original Array: ",arr)
print("Reversed Array: ",reverse)
```

#### b) Find the most frequent value and their indice(s) in the following arrays: 
#### i. x = np.array([1,2,3,4,5,1,2,1,1,1]) 

```python
x = np.array([1,2,3,4,5,1,2,1,1,1])

values,counts = np.unique(x,return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(x==most_frequent)[0]

print("Most Frequent Value: ",most_frequent)
print("Indices: ",indices)
```

#### ii. y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(y == most_frequent)[0]

print("Most frequent value:", most_frequent)
print("Indices:", indices)
```

### Q4. Write program to create an 1-D NumPy array named << Your Name >> with evenly spaced 25 numbers from 10 to 100 using linspace(). Print the dimensions of the array, shape, total elements, the data type of each element and total number of bytes consumed by the array. Find the transpose of this array using reshape() attribute. Can we do the same with T attribute? 

```python
Anwesha = np.linspace(10,100,25)
print("Array: ")
print(Anwesha)
```

```python
print("Dimensions: ",Anwesha.ndim)
```

```python
print("Shape: ",Anwesha.shape)
```

```python
print("Total Elements: ",Anwesha.shape)
```

```python
print("Data type:", Anwesha.dtype)
```

```python
print("Total bytes:", Anwesha.nbytes)
```

```python
transpose_array = Anwesha.reshape(25, 1)

print("Transpose using reshape():")
print(transpose_array)
```

```python
print("Transpose using T:")
print(Anwesha.T)   # A 1-D NumPy array cannot actually be transposed using .T.
```

### Q5. Create a 2-D Array of three rows and four columns, named ucs420_<< your_name >> with following values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35.

```python
ucs420_Anwesha = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])
```

#### Compute the mean, median, max, min, unique elements. 

```python
print("Original Array: \n",ucs420_Anwesha)
```

```python
print("Mean: ",np.mean(ucs420_Anwesha))
```

```python
print("Median: ",np.median(ucs420_Anwesha))
```

```python
print("Maximum: ",np.max(ucs420_Anwesha))
```

```python
print("Minimum: ",np.min(ucs420_Anwesha))
```

```python
print("Unique: ",np.unique(ucs420_Anwesha))
```

#### Reshape the array to four rows and three columns and name it as reshaped_ ucs420_<< your_name >>. 

```python
reshaped_ucs420_Anwesha = ucs420_Anwesha.reshape(4,3)
print("Reshaped Array (4 x 3):")
print(reshaped_ucs420_Anwesha)
```

#### Resize the array to two rows and three columns and name it as resized_ ucs420_<< your_name >>. 

```python
resized_ucs420_Anwesha = np.resize(ucs420_Anwesha, (2, 3))
print("Resized Array (2 x 3):")
print(resized_ucs420_Anwesha)
```
